## Experiment-1: Data Analysis


### 1.1: Read the employee dataset and display the average salary of each department using a bar chart.

```python
# Importing the neccessary libraries:
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Loading the dataset:
employee_data = pd.read_csv('employee_information_100.csv')
employee_data.head()
```

```python
# Calculating average salary of each department:
avg_salary = employee_data.groupby('Department')['Salary'].mean().reset_index()
print(avg_salary)
```

```python
plt.bar(avg_salary['Department'], avg_salary['Salary'])
plt.xlabel('Department')
plt.ylabel('Average Salary')
plt.title('Average Salary of Each Department')
plt.show()
```

### Experiment 1.2: Display the number of employees working in each department.

```python
num_employees = employee_data['Department'].value_counts().reset_index()
num_employees.columns = ['Department', 'Number of Employees']
print(num_employees)
```

### Experiment 1.3: Display the percentage of male and female employees using a pie chart.

```python
# Displaying the number of male and female employees using a pie chart:
gender_counts = employee_data['Gender'].value_counts()
plt.pie(gender_counts, labels=gender_counts.index, autopct='%1.1f%%')
plt.title('Percentage of Male and Female Employees')
plt.show()
```

### Experiment 1.4: Draw a histogram showing salary distribution.


```python
# Drawing a histogram to show salary distribution:
plt.hist(employee_data['Salary'], bins=10, edgecolor = 'black')
plt.xlabel('Salary')
plt.ylabel('Frequency')
plt.title('Salary Distribution')
plt.show()
```

### Experiment 1.5: Draw a scatter plot showing the relationship between experience and salary.

```python
# Scatter plot to show relationship between experience and salary:
plt.scatter(employee_data['Experience_Years'], employee_data['Salary'])
plt.xlabel('Experience')
plt.ylabel('Salary')
plt.title('Experience vs. Salary')
plt.show()
```

### Experiment 1.6: Display the top 10 employees with the highest salary.

```python
# Displaying the top 10 employees with the highest salary:
top_10_employees = employee_data.nlargest(10, 'Salary')
print(top_10_employees[['Name', 'Salary']])
```

### Experiment 1.7: Experiment 1.7: Find the highest salary in every department.


```python
# Displaying highest salary in every department:
highest_salary_dept = employee_data.groupby('Department')['Salary'].max().reset_index()
print(highest_salary_dept)
```

### Experiment 1.8: Display employees whose salary is greater than the overall average salary.

```python
# Displaying employees whose salary is greater than overall average salary:
overall_avg_salary = employee_data['Salary'].mean()
high_salary_employees = employee_data[employee_data['Salary'] > overall_avg_salary]
print(high_salary_employees[['Name', 'Salary']])
```

### Experiment 1.9: Calculate the average years of experience in each department.

```python
# Calculating average years of experience in each department:
avg_experience_dept = employee_data.groupby('Department')['Experience_Years'].mean().reset_index()
print(avg_experience_dept)
```

### Experiment 1.10: Display the distribution of employee ages using a histogram.

```python
# Distribution of employee ages using histogram:
plt.hist(employee_data['Age'], bins=10, edgecolor='black')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.title('Distribution of Employee Ages')
plt.show()
```
