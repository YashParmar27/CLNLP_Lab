## Experiment-3: Stemming, Lemmatization and Regular Expression

#### Experiment-3.1: Write a Python program to perform stemming using the Porter Stemmer. Apply stemming to the following words. Display the original words and their stems.

```python
import nltk
from nltk.stem import PorterStemmer

word_list = ["playing", "played", "plays", "studies", "studying", "connected", "connection", "computers"]

stemmer = PorterStemmer()

for word in word_list:
    print(word, "->", stemmer.stem(word))
```

#### Experiment-3.2: Write a Python program to perform lemmatization on a given input words.

```python
from nltk.stem import WordNetLemmatizer

nltk.download("wordnet")
nltk.download("omw-1.4")

words = [
    ("cats", "n"),
    ("dogs", "n"),
    ("running", "v"),
    ("runs", "v"),
    ("ran", "v"),
    ("studies", "v"),
    ("studying", "v"),
    ("better", "a"),
    ("children", "n"),
    ("mice", "n"),
    ("went", "v"),
    ("ate", "v"),
    ("leaves", "n"),
    ("caring", "v")
]

lemmatizer = WordNetLemmatizer()

for word, pos in words:
    print(word, "->", lemmatizer.lemmatize(word, pos=pos))
```

#### Experiment-3.3: Write a Python program using Regular Expressions (Regex) to extract and process the following information from a text.

```python
# Regex for Email Addresses:
import re

text = """Welcome to the Natural Language Processing workshop! For more information,
          visit https://www.nlpworkshop.com or www.python.org. You can contact the
          coordinator at nlpworkshop@gmail.com or support@python.org. For registration,
          call +91-9876543210 or 9123456789. Follow us on social media @NLPWorkshop and
          @PythonLearner. Share your experience using #NLP, #Python, and #MachineLearning.
          You can also visit https://github.com/NLPWorkshop for the latest updates."""

emails = re.findall(r'\S+@\S+', text)
print("Emails:", emails)
```

```python
# Regex for URLs:
urls = re.findall(r'https?://\S+|www\.\S+', text)

print("URLs:", urls)
```

```python
# Regex for mobile numbers:
mobiles = re.findall(r'\+91-\d{10}|\b\d{10}\b', text)

print("Mobile Numbers:", mobiles)
```

```python
# Regex for hashtags:
hashtags = re.findall(r'#\w+', text)

print("Hashtags:", hashtags)
```

```python
# Regex for mentions:
mentions = re.findall(r'@\w+', text)

print("Mentions:", mentions)
```
