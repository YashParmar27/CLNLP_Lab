#### 1. Convert text to lowercase. (Also display how many uppercase?)

```python
# Reading the text and converting text to lowercase, displaying number of uppercase characters:
with open('/content/drive/MyDrive/CLNLP_Lab_Data/2.1_text_data.txt', 'r') as file:
    text = file.read()

uppercase_count = sum(1 for char in text if char.isupper())
text = text.lower()

print("Text in lowercase:", text)
print("Number of uppercase characters:", uppercase_count)
```

#### 2. Remove punctuation.        (Also display how many different types of punctuations?)

```python
punctuations = set(c for c in text if c in string.punctuation)
text = ''.join(c for c in text if c not in string.punctuation)

print(text)
print("Different types of punctuation:", len(punctuations))
```

#### 3. Remove numbers.            (Also display how many numbers?)

```python
numbers = sum(1 for c in text if c.isdigit())
text = ''.join(c for c in text if not c.isdigit())

print(text)
print("Number of numbers:", numbers)
```

#### 4. Remove extra whitespace. (How much extra whitespace is present?)

```python
words = text.split()
extra_whitespace = len(text) - len(' '.join(words))

text = ' '.join(words)

print(text)
print("Extra whitespace:", extra_whitespace)
```

#### 5. Display the cleaned text.    

```python
print(text)
```

### 2.2: Write a Python program to perform word tokenization and sentence tokenization on a given text. Display the resulting words and sentences separately.


#### 1. Write a Python program to perform word tokenization and sentence tokenization on a given text. Display the resulting words and sentences separately. (Three separate code: using NLTK, using spaCy  and without using NLTK/spacy)

#### 1. NLTK

```python
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize

nltk.download('punkt_tab')

text = "Hello everyone. This is a NLP lab. We are learning tokenization."

words = word_tokenize(text)
sentences = sent_tokenize(text)

print("Words:", words)
print("Sentences:", sentences)
```

#### 2. spaCy

```python
import spacy

nlp = spacy.load("en_core_web_sm")

text = "Hello everyone. This is a NLP lab. We are learning tokenization."

doc = nlp(text)

words = [token.text for token in doc]
sentences = [sentence.text for sentence in doc.sents]

print("Words:", words)
print("Sentences:", sentences)
```

#### 3. without NLTK/ spaCy

```python
text = "Hello everyone. This is a NLP lab. We are learning tokenization."

words = text.split()
sentences = text.split('.')

print("Words:", words)
print("Sentences:", [s.strip() for s in sentences if s.strip()])
```

#### 2.3: Write a Python program to tokenize a text and remove commonly occurring stop words such as is, the, a, an, in, of, and, to, etc. Display the original and filtered tokens. (Also, display Stop Words Separately).


```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

nltk.download('stopwords')

with open('/content/drive/MyDrive/CLNLP_Lab_Data/2.3_clean_data.txt', 'r') as file:
    text = file.read()
tokens = word_tokenize(text)
stop_words = set(stopwords.words('english'))

filtered = [word for word in tokens if word.lower() not in stop_words]
removed = [word for word in tokens if word.lower() in stop_words]

print("Original Tokens:", tokens)
print("Filtered Tokens:", filtered)
print("Stop Words:", removed)
```
