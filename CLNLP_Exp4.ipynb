#### Calculate the frequency of each word and save the output in CSV format (Term, Frequency):

```python
import nltk
import csv
from collections import Counter
from nltk.tokenize import word_tokenize
from nltk import ne_chunk, pos_tag
from nltk.tree import Tree

nltk.download("punkt_tab")
nltk.download("averaged_perceptron_tagger_eng")
nltk.download("maxent_ne_chunker_tab")
nltk.download("words")

# Load the document
with open("/content/drive/MyDrive/4.1_4 (1).2_input.txt", "r", encoding="utf-8") as file:
  text = file.read()

# Tokenize words
words = word_tokenize(text.lower())
words = [word for word in words if word.isalnum()]

# Count frequency
frequency = Counter(words)

# Save frequency to CSV
with open("term_frequency_4_1.csv", "w", newline="", encoding="utf-8") as file:
  writer = csv.writer(file)
  writer.writerow(["Term", "Frequency"])

  for word, count in frequency.items():
    writer.writerow([word, count])

# Display top 10 terms
print("Top 10 terms:")

for word, count in frequency.most_common(10):
  print(word, count)
```

```python
# Named-Enitity Recognition:
import spacy

nlp = spacy.load("en_core_web_sm")
doc = nlp(text)

print("Named Entities:")
for entity in doc.ents:
  print(entity.text, "->", entity.label_)
```

#### Experiment-4.2: To perform Term-Frequency Analysis on a given text without using any toolkit.

```python
import csv
import re
# Load the document
with open("/content/drive/MyDrive/4.1_4 (1).2_input.txt", "r", encoding="utf-8") as file:
  text = file.read()

# Convert to lowercase
text = text.lower()

# Remove punctuation
text = re.sub(r'[^\w\s]', '', text)

# Tokenize
words = text.split()

# Calculate frequency
frequency = {}
for word in words:
  if word in frequency:
    frequency[word] += 1
  else:
    frequency[word] = 1

# Save frequency to CSV
with open("term_frequency_4_2.csv", "w", newline="", encoding="utf-8") as file:
  writer = csv.writer(file)
  writer.writerow(["Term", "Frequency"])

  for word, count in frequency.items():
    writer.writerow([word, count])

# Display top 10 terms
top10 = sorted(frequency.items(), key=lambda x: x[1], reverse=True)[:10]

print("Top 10 terms:")
for word, count in top10:
  print(word, count)
```

#### Experiment-4.3: Write a Python program without using NLTK, spaCy, or any other NLP toolkit.

```python
# Read multiple text documents:
import re
import math

# Load multiple documents
documents = [ "/content/drive/MyDrive/doc_1.txt",
              "/content/drive/MyDrive/doc_2.txt",
              "/content/drive/MyDrive/doc_3.txt" ]

texts = []
for file_name in documents:
  with open(file_name, "r", encoding="utf-8") as file:
    text = file.read().lower()
    text = re.sub(r'[^\w\s]', '', text)
    texts.append(text.split())

# Calculate TF
tf = []
for words in texts:

  count = {}

  for word in words:
    count[word] = count.get(word, 0) + 1

  total = len(words)

  doc_tf = {}

  for word in count:
    doc_tf[word] = count[word] / total

  tf.append(doc_tf)

# Calculate DF
df = {}
for words in texts:
  for word in set(words):
    df[word] = df.get(word, 0) + 1

# Calculate IDF
N = len(texts)

idf = {}
for word in df:
  idf[word] = math.log(N / df[word])

# Calculate TF-IDF
tfidf = []
for i in range(N):
  scores = {}

  for word in tf[i]:
    scores[word] = tf[i][word] * idf[word]

  tfidf.append(scores)

# Display TF, DF, IDF and TF-IDF

for i in range(N):
    print("\nDocument", i + 1)
    print(f"{'Term':<20}{'TF':<10}{'DF':<10}{'IDF':<10}{'TF-IDF':<10}")
    print("-" * 60)

    for word in tf[i]:
        print(
            f"{word:<20}"
            f"{tf[i][word]:<10.4f}"
            f"{df[word]:<10}"
            f"{idf[word]:<10.4f}"
            f"{tfidf[i][word]:<10.4f}"
        )

# Display top 10 TF-IDF terms

for i in range(N):
    top10 = sorted(
        tfidf[i].items(),
        key=lambda x: x[1],
        reverse=True
    )[:10]

    print("\nTop 10 TF-IDF terms - Document", i + 1)
    print(f"{'Term':<20}{'TF-IDF':<10}")
    print("-" * 30)

    for word, score in top10:
        print(f"{word:<20}{score:<10.4f}")
```
